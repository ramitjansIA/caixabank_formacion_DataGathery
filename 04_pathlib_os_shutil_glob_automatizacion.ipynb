## 7. Buscar archivos con `glob`

````python
pdfs_no_normalizados = list(
    entrada.glob(
        "*.pdf"
    )
)

[
    ruta.name
    for ruta in pdfs_no_normalizados
]
````

### Problema visible

`glob("*.pdf")` no encuentra `factura_002.PDF`.

Esto demuestra que los patrones pueden ser sensibles al formato real de los nombres.

## 7.1 Solución robusta mediante inspección

````python
pdfs = [
    ruta
    for ruta in entrada.iterdir()
    if (
        ruta.is_file()
        and ruta.suffix.lower()
        == ".pdf"
    )
]

[
    ruta.name
    for ruta in pdfs
]
````

## 8. Búsqueda recursiva con `rglob`

````python
todos_los_pdf = [
    ruta
    for ruta in entrada.rglob("*")
    if (
        ruta.is_file()
        and ruta.suffix.lower()
        == ".pdf"
    )
]

[
    str(
        ruta.relative_to(
            raiz
        )
    )
    for ruta in todos_los_pdf
]
````

### Interpretación

`rglob` permite buscar en todas las subcarpetas.

Es fundamental para automatizaciones documentales recursivas.

## 9. Buscar por múltiples extensiones

````python
extensiones_documentos = {
    ".pdf",
    ".xlsx",
    ".csv",
    ".txt"
}

documentos = [
    ruta
    for ruta in entrada.rglob("*")
    if (
        ruta.is_file()
        and ruta.suffix.lower()
        in extensiones_documentos
    )
]

[
    ruta.name
    for ruta in documentos
]
````

## 12. Copiar un archivo con `shutil.copy2`

````python
origen = (
    raiz
    / "entrada"
    / "factura_001.pdf"
)

destino = (
    raiz
    / "salida"
    / "PDF"
    / "factura_001.pdf"
)

shutil.copy2(
    origen,
    destino
)

print(
    "Origen existe:",
    origen.exists()
)

print(
    "Copia existe:",
    destino.exists()
)
````

### Interpretación

`copy2` conserva más metadatos que `copy`.

El archivo original sigue existiendo.

## 13. Copiar una carpeta completa

````python
origen_carpeta = (
    raiz
    / "entrada"
    / "clientes"
)

destino_carpeta = (
    raiz
    / "backup"
    / "clientes"
)

shutil.copytree(
    origen_carpeta,
    destino_carpeta
)

print(
    destino_carpeta.exists()
)
````

## 13.1 Provocar colisión de carpeta

````python
try:
    shutil.copytree(
        origen_carpeta,
        destino_carpeta
    )

except FileExistsError as error:
    print(
        type(error).__name__,
        "->",
        error
    )
````

### Interpretación

La segunda copia falla porque el destino ya existe.

Una automatización debe decidir qué política quiere:

- fallar;
- sobrescribir;
- fusionar;
- generar otro nombre.

## 14. Copiar permitiendo destino existente

````python
shutil.copytree(
    origen_carpeta,
    destino_carpeta,
    dirs_exist_ok=True
)

print(
    "Copia actualizada."
)
````

## 15. Mover archivos

````python
archivo_mover = (
    raiz
    / "entrada"
    / "ventas.csv"
)

destino_mover = (
    raiz
    / "salida"
    / "Datos"
    / "ventas.csv"
)

shutil.move(
    str(
        archivo_mover
    ),
    str(
        destino_mover
    )
)

print(
    "Origen:",
    archivo_mover.exists()
)

print(
    "Destino:",
    destino_mover.exists()
)
````

### Diferencia conceptual

**Copiar** conserva el origen.

**Mover** cambia la ubicación.

## 17. Provocar colisión de nombres

````python
origen_colision = (
    raiz
    / "entrada"
    / "factura_001.pdf"
)

destino_colision = (
    raiz
    / "salida"
    / "PDF"
    / "factura_001.pdf"
)

print(
    "Destino ya existe:",
    destino_colision.exists()
)
````

## 17.1 Estrategia segura: generar nombre alternativo

````python
def generar_destino_unico(
    destino
):
    destino = Path(
        destino
    )

    if not destino.exists():
        return destino

    contador = 1

    while True:
        candidato = (
            destino
            .with_name(
                f"{destino.stem}_{contador}{destino.suffix}"
            )
        )

        if not candidato.exists():
            return candidato

        contador += 1


destino_unico = (
    generar_destino_unico(
        destino_colision
    )
)

print(
    destino_unico.name
)
````

## 17.2 Copiar sin sobrescribir

````python
shutil.copy2(
    origen_colision,
    destino_unico
)

print(
    destino_unico.exists()
)
````

### Interpretación

No sobrescribimos silenciosamente.

Generamos una alternativa controlada.

## 18. Obtener metadatos de archivos

````python
def inspeccionar_archivo(
    ruta
):
    ruta = Path(
        ruta
    )

    stat = ruta.stat()

    return {
        "nombre":
            ruta.name,

        "extension":
            (
                ruta.suffix.lower()
                if ruta.suffix
                else "sin_extension"
            ),

        "bytes":
            stat.st_size,

        "carpeta":
            str(
                ruta.parent
            )
    }


inspeccionar_archivo(
    raiz
    / "entrada"
    / "factura_001.pdf"
)
````

## 18.1 Crear inventario recursivo

````python
inventario = pd.DataFrame(
    [
        inspeccionar_archivo(
            ruta
        )
        for ruta
        in raiz.rglob("*")
        if ruta.is_file()
    ]
)

inventario.head(
    10
)
````

## 18.2 Visualizar tamaños

````python
tamano_extension = (
    inventario
    .groupby(
        "extension",
        as_index=False
    )[
        "bytes"
    ]
    .sum()
)

plt.figure(
    figsize=(8, 5)
)

plt.bar(
    tamano_extension[
        "extension"
    ],
    tamano_extension[
        "bytes"
    ]
)

plt.ylabel(
    "Bytes"
)

plt.title(
    "Tamaño acumulado por extensión"
)

plt.tight_layout()
plt.show()
````

## 19. Crear una función de clasificación física

````python
REGLAS = {
    ".pdf": "PDF",
    ".xlsx": "Excel",
    ".csv": "Datos",
    ".jpg": "Imagenes",
    ".jpeg": "Imagenes",
    ".png": "Imagenes"
}


def categoria_destino(
    ruta
):
    ruta = Path(
        ruta
    )

    if not ruta.suffix:
        return "Revision"

    return REGLAS.get(
        ruta.suffix.lower(),
        "Revision"
    )


for archivo in (
    raiz
    / "entrada"
).iterdir():

    if archivo.is_file():
        print(
            archivo.name,
            "->",
            categoria_destino(
                archivo
            )
        )
````

## 20. Simular organización sin mover

````python
plan = []

for archivo in (
    raiz
    / "entrada"
).iterdir():

    if not archivo.is_file():
        continue

    categoria = (
        categoria_destino(
            archivo
        )
    )

    destino = (
        raiz
        / "salida"
        / categoria
        / archivo.name
    )

    plan.append({
        "origen":
            str(
                archivo
            ),

        "categoria":
            categoria,

        "destino":
            str(
                destino
            ),

        "colision":
            destino.exists()
    })


plan_df = pd.DataFrame(
    plan
)

plan_df
````

### Regla profesional

Antes de ejecutar operaciones destructivas es muy útil generar un **plan**.

Así podemos revisar:

- destinos;
- colisiones;
- elementos desconocidos.

## 20.1 Visualizar colisiones

````python
plan_df[
    [
        "categoria",
        "colision"
    ]
].value_counts()
````

## 21. Ejecutar plan de forma segura

````python
def ejecutar_plan(
    plan,
    mover=False
):
    resultados = []

    for item in plan:

        origen = Path(
            item["origen"]
        )

        destino = Path(
            item["destino"]
        )

        destino.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        destino_final = (
            generar_destino_unico(
                destino
            )
        )

        if mover:
            shutil.move(
                str(
                    origen
                ),
                str(
                    destino_final
                )
            )

            operacion = (
                "movido"
            )

        else:
            shutil.copy2(
                origen,
                destino_final
            )

            operacion = (
                "copiado"
            )

        resultados.append({
            "origen":
                str(
                    origen
                ),

            "destino":
                str(
                    destino_final
                ),

            "operacion":
                operacion
        })

    return pd.DataFrame(
        resultados
    )


resultado_plan = ejecutar_plan(
    plan,
    mover=False
)

resultado_plan
````

## 21.1 Validar ejecución

````python
assert (
    len(
        resultado_plan
    )
    == len(
        plan
    )
)

assert all(
    Path(
        destino
    ).exists()
    for destino
    in resultado_plan[
        "destino"
    ]
)

print(
    "Plan ejecutado correctamente."
)
````

## 25. Validación antes de borrar

````python
def eliminar_directorio_seguro(
    ruta,
    raiz_permitida
):
    ruta = Path(
        ruta
    ).resolve()

    raiz_permitida = Path(
        raiz_permitida
    ).resolve()

    if raiz_permitida not in ruta.parents:
        raise ValueError(
            "La ruta está fuera de la raíz permitida."
        )

    if not ruta.exists():
        raise FileNotFoundError(
            ruta
        )

    shutil.rmtree(
        ruta
    )


prueba_segura = (
    raiz
    / "borrar_demo"
)

prueba_segura.mkdir()

eliminar_directorio_seguro(
    prueba_segura,
    raiz
)

print(
    prueba_segura.exists()
)
````
